## 清空 GPU 記憶體

有時您需要清空 GPU 記憶體，可能是為了在實驗出錯時重設 GPU 狀態，或者在不同筆記本之間需要為新的一組練習提供全新環境。

清空 GPU 記憶體主要有 3 種方法。第一種是使用筆記本正上方選單中的 ⟳ 按鈕。現在點擊它，您應該會看到一個提示，詢問是否可以重設 GPU 記憶體。

第二種方法是通過我們將在每個Notebook結尾提供的程式碼區塊。以下是其中一個區塊，您可以執行它來養成正確的習慣。

```python
import IPython
app = IPython.Application.instance()
app.kernel.do_shutdown(True)
```
